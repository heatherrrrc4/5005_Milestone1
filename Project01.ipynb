```python
import json
import requests

url = "https://raw.githubusercontent.com/heatherrrrc4/5005_Milestone1/refs/heads/main/WatchHistory.json"

response = requests.get(url)
data = response.json()

print("Total records:", len(data))
print("First item:", data[0])
```

## Show Data

Show some interesting findings.

```python
# This visualization shows the distribution of different types of content
# in my YouTube watch history based on keywords such as "cozy", "sleep", and "music".
# But others are drama shows I watched
```

```python
import matplotlib.pyplot as plt

cozy = 0
music = 0
other = 0

for item in data:
    title = item["title"].lower()
    if "sleep" in title or "cozy" in title or "rain" in title:
        cozy += 1
    elif "music" in title or "jazz" in title:
        music += 1
    else:
        other += 1

labels = ["Cozy/Sleep", "Music", "Other"]
values = [cozy, music, other]

plt.figure(figsize=(6,4))
plt.bar(labels, values)
plt.title("Distribution of My YouTube Viewing Content")
plt.xlabel("Category")
plt.ylabel("Count")

for i, v in enumerate(values):
    plt.text(i, v, str(v), ha='center')  # showing numbers

plt.show()
```
